```python
def function_example(*args, **kwargs):
    for arg in args:
        print(arg)
    for key, value in kwargs.items():
        print(f"{key} = {value}")

function_example(1, 2, 3, a=4, b=5)
```

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1
count_up_to(5)
```

```python
import threading

def print_numbers():
    for i in range(1, 6):
        print(i)

def print_letters():
    for letter in 'abcde':
        print(letter)

t1 = threading.Thread(target=print_numbers)
t2 = threading.Thread(target=print_letters)

t1.start()
t2.start()

t1.join()
t2.join()
```

```python
# Creating a class named 'Robot' at runtime
def greet(self):
    print(f"Hello, I am {self.name}")

# Dynamic creation
Robot = type('Robot', (object,), {'name': 'R2-D2', 'say_hello': greet})

# Using the dynamic class
my_robot = Robot()
my_robot.say_hello()  # Output: Hello, I am R2-D2
```

```python
import time

def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)  # Executing the original function
        end_time = time.time()
        print(f"Execution time: {end_time - start_time} seconds")
        return result
    return wrapper

@timer_decorator
def heavy_computation():
    time.sleep(2) # even without this decorator is being used, this just acts as placeholder
    print("Finished computing.")

heavy_computation()
```

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def diameter(self):
        return self._radius * 2

circle = Circle(5)
print(circle.diameter)  # Output: 10
```

```python
class Descriptor:
    def __get__(self, instance, owner):
        return instance._value

    def __set__(self, instance, value):
        instance._value = value.upper()

class MyClass:
    attribute = Descriptor()

    def __init__(self, value):
        self._value = value

obj = MyClass('hello')
print(obj.attribute)  # hello
obj.attribute = 'world' # descriptor object is called, Descriptor.__set__(self, instance, value)
print(obj.attribute)  # WORLD
```

```python
class Person:
    def __init__(self, name):
        self.name = name
    
    def __str__(self):
        return f"Person named {self.name}"

p = Person("Alice")
print(p)  # Output: "Person named Alice"
```
