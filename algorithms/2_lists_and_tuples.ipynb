```python
list1 = ['apple','orange','banana','peach','grapes']
list1
```

**Adding Elements into List**

We can add elements to a list using the following methods:

* append(): Adds an element at the end of the list.
* extend(): Adds multiple elements to the end of the list.
* insert(): Adds an element at a specific position.
* clear(): removes all items.

```python
list1.append('kiwi')
print(list1)
```

```python
list1.insert(2,'strawberry')
print(list1)
```

```python
list1[1:3] = ['blackcurrent','watermelon'] # here we are replacing the values within the list
print(list1)
```

```python
a = []

a.append(10)  
print("After append(10):", a)  

a.insert(0, 5)
print("After insert(0, 5):", a) 

a.extend([15, 20, 25])  
print("After extend([15, 20, 25]):", a) 

a.clear()
print("After clear():", a)
```

**Removing Elements from List**

We can remove elements from a list using:

remove(): Removes the first occurrence of an element.
pop(): Removes the element at a specific index or the last element if no index is specified.
del statement: Deletes an element at a specified index.

```python
a = [10, 20, 30, 40, 50]

a.remove(30)  
print("After remove(30):", a)

popped_val = a.pop(1)  
print("Popped element:", popped_val)
print("After pop(1):", a) 

del a[0]  
print("After del a[0]:", a)
```

```python
#some operations on strings
def trim(str):
    return str.strip()
def titleIt(str):
    return str.title()
def swap(str):
    return str.swapcase()
def exists(str,x):
    return str.find(x)
print(trim("hello"))
print(titleIt("hello"))
print(swap("hello"))
print(exists("hello","hel"))
```

```python
#convert the binary number to decimal
def binarytoDecimal(b):
    val = 0
    lastdigit = 0
    power = 0
    while b > 0:
        lastdigit = b % 10
        val += lastdigit * 2**power
        power += 1
        b //=10
    return val
binarytoDecimal(111)
```

**Tuples** are ordered immutable collection of items, once its defined it cannot be changed

```python
tuple1 = ('emerald', 'ametheyst','peridot')
print(tuple1)
print(len(tuple1))
print(type(tuple1))
```

```python
tuple2 = tuple(['garnet', 'opal','ruby','topaz'])
print(type(tuple2))
print(tuple2)
```

```python
print(tuple1[-1]) # returns the last element
tuple3 = tuple1 + tuple2
print(tuple3)
tuple4 = tuple1 * 2
print(tuple4)
```

**Python Dictionary**

Python dictionary is a data structure that stores the value in key: value pairs. Values in a dictionary can be of any data type and can be duplicated, whereas keys can't be repeated and must be immutable.

* Keys are case sensitive which means same name but different cases of Key will be treated distinctly.
* Keys must be immutable which means keys can be strings, numbers or tuples but not lists.
* Duplicate keys are not allowed and any duplicate key will overwrite the previous value.
* Internally uses hashing. Hence, operations like search, insert, delete can be performed in Constant Time.
* From Python 3.7 Version onward, Python dictionary are Ordered.
